## **Ejercicio 1: Carrito de compras**

Crea una clase llamada **`CarritoCompras`** que modele un carrito de compras de una tienda online.

### **La clase debe tener:**

**Atributos:**
- `cliente`: nombre del dueño del carrito  
- `productos`: diccionario donde la clave es el nombre del producto y el valor es su precio  

**Métodos:**
- `agregar(producto, precio)`: añade un producto al carrito  
- `total()`: devuelve la suma de los precios de todos los productos  
- `__str__`: retorna una representación legible del carrito (cliente, productos y total)  
- `__len__`: devuelve la cantidad de productos en el carrito  

---

### **Adicionalmente:**
- Crear al menos **dos instancias** de `CarritoCompras`  
- Agregar productos a cada una  
- Imprimir los carritos usando `print()`  
- Mostrar la cantidad de productos utilizando `len()`  

```python
# -- DEFINICIO DE LA CLASE CARRITOCOMPRAS
class CarritoCompras:
    # -- CONSTRUCTOR INICIAL PARA EL CARRITO, ARRCANA CON UN CLIENTE Y UN DICCIONARIO VACIO DE PRODUCTOS
    def __init__(self, cliente):
        self.cliente = cliente
        self.productos = {}

    # -- FUNCION DE AGREGAR, AGREGA UN PRODUCTO Y SU PRECIO
    def agregar(self, producto, precio):
        self.productos[producto] = precio

    # -- FUNCION TOTAL PARA CALCULAR LOS PRECIOS
    def total(self):
        return sum(self.productos.values())

    # -- FUNCION PARA EL RESULTADO FINAL (VISUAL TIPO FACTURA AUNQUE SEA EN CONSOLA JAJA)
    def __str__(self):
        ancho = 45
        separador1 = "=" * ancho
        separador2 = "-" * ancho

        recibo = f"\n{separador1}\n"
        recibo += f" TIENDA - HOY NO FIO, MAÑANA SI! \n MAÑANA VUELVE A LEER EL MENSAJE\n"
        recibo += f"{separador1}\n"
        recibo += f"Cliente: {self.cliente}\n"
        recibo += f"{separador2}\n"

        for prod, precio in self.productos.items():
            precio_fmt = f"{precio:,.0f}".replace(",", ".")
            recibo += f"{prod[:20]:<20} $ {precio_fmt:>10}\n"

        recibo += f"{separador2}\n"

        # TOTAL FINAL DEL CARRITO
        total_fmt = f"{self.total():,.0f}".replace(",", ".")
        recibo += f"TOTAL A PAGAR:       $ {total_fmt:>10} COP\n"
        recibo += f"{separador1}\n"

        return recibo

    # -- FUNCION PARA MOSTRAR LA CANTDAD DE PRODUCTOS
    def __len__(self):
        return len(self.productos)

# --- INSTANCIAS ---

# INSTANCIA 1
carrito_tech = CarritoCompras("Jose David")
carrito_tech.agregar("iPhone 15 Pro", 4800000)
carrito_tech.agregar("Funda Silicona", 120000)
carrito_tech.agregar("Cargador 20W", 95000)

# INSTANCIA 2
carrito_mercado = CarritoCompras("Daniela")
carrito_mercado.agregar("Arroz 5kg", 22000)
carrito_mercado.agregar("Aceite Girasol", 18500)
carrito_mercado.agregar("Café Sello Rojo", 15000)
carrito_mercado.agregar("Bolsa de Leche x6", 28000)

# INSTANCIA 3
carrito_hogar = CarritoCompras("Algún RandoMas")
carrito_hogar.agregar("Silla Ergonómica", 550000)
carrito_hogar.agregar("Lámpara LED", 85000)

# -- IMPRESION DE RESULTADOS (AQUI PUEDE EJECUTARSE UNA O TODAS)
instancias = [carrito_tech, carrito_mercado, carrito_hogar]

# -- PRIN FINAL Y EJECUCIONES
for i, carrito in enumerate(instancias, 1):
    print(f"MOSTRANDO INSTANCIA #{i}:")
    print(carrito)
    print(f"Resumen: {len(carrito)} productos registrados.\n")
```

## **Ejercicio 2: Escritura de datos en un archivo CSV**

```python
import csv

# -- Definición de los datos
frutas = [
    ["Manzana", "Roja", "Dulce"],
    ["Plátano", "Amarillo", "Dulce"],
    ["Lima", "Verde", "Ácida"]
]

# -- Nombre del archivo
# -- el archivo se crea en la carpeta temporal de este colab
archivo_csv = "frutas.csv"

# -- Escritura de los datos
with open(archivo_csv, mode="w", encoding="utf-8", newline="") as file:
    escritor = csv.writer(file)

    # Escribimos todas las filas de una vez
    escritor.writerows(frutas)

print(f"El archivo '{archivo_csv}' ha sido creado exitosamente con {len(frutas)} filas.")
```

## **Ejercicio 3: Lectura de datos de un archivo CSV**

```python
import csv

# -- GENERA EL ARREGLO VACIO DE LOS DATOS DE LAS FRUTAS
datos_frutas = []

# -- CONSULTA EL ARCHIVO TEMPORAL CREADO EN LA EJECUCIÓN ANTERIOR
with open("frutas.csv", mode="r", encoding="utf-8") as archivo:
    lector = csv.reader(archivo)

    for fila in lector:
        datos_frutas.append(fila)
# -- IMPRESIÓN DE LOS DATOS Y LOS RESULTADOS
print(datos_frutas)
```

## **Ejercicio 4: Procesamiento y análisis de datos desde una API**

```python
import requests
import pandas as pd
import os

# 1. EXTRACCIÓN

base_url = "https://jsonplaceholder.typicode.com"

todos = requests.get(f"{base_url}/todos").json()
users = requests.get(f"{base_url}/users").json()

# 2. CARGA A DATAFRAMES

df_todos = pd.DataFrame(todos)
df_users = pd.DataFrame(users)

# 3. TRANSFORMACIÓN

df_users = df_users.rename(columns={
    "id": "userId",
    "name": "nombre_usuario",
    "email": "email_usuario"
})

df_pipeline = df_todos.merge(
    df_users[["userId", "nombre_usuario", "email_usuario"]],
    on="userId",
    how="left"
)

df_pipeline["estado"] = df_pipeline["completed"].apply(
    lambda x: "Completada" if x else "Pendiente"
)

df_pipeline = df_pipeline[[
    "id",
    "userId",
    "nombre_usuario",
    "email_usuario",
    "title",
    "estado"
]]

# 4. ANÁLISIS

resumen_usuarios = (
    df_pipeline
    .groupby(["userId", "nombre_usuario", "estado"])
    .size()
    .reset_index(name="cantidad_tareas")
)

resumen_final = (
    resumen_usuarios
    .pivot_table(
        index=["userId", "nombre_usuario"],
        columns="estado",
        values="cantidad_tareas",
        fill_value=0
    )
    .reset_index()
)

resumen_final["total_tareas"] = (
    resumen_final["Completada"] + resumen_final["Pendiente"]
)

resumen_final["porcentaje_completado"] = (
    resumen_final["Completada"] / resumen_final["total_tareas"] * 100
).round(2)

# 5. EXPORTACIÓN EN COLAB

carpeta = "/content/pipeline_jsonplaceholder_todos"
os.makedirs(carpeta, exist_ok=True)

df_pipeline.to_csv(f"{carpeta}/tareas_procesadas.csv", index=False, encoding="utf-8")
resumen_final.to_csv(f"{carpeta}/resumen_tareas_usuarios.csv", index=False, encoding="utf-8")

# 6. RESULTADOS

print("Pipeline ejecutada correctamente")

print("\nTareas procesadas:")
print(df_pipeline.head())

print("\nResumen por usuario:")
print(resumen_final)

print(f"\nArchivos creados en: {carpeta}")
```
